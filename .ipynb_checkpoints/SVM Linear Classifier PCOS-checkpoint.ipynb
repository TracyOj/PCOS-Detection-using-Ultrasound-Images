```python
# Importlibraries
import os
import cv2
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score 
from sklearn.neighbors import KNeighborsClassifier
from skimage.transform import resize 
from skimage.io import imread 
import matplotlib.pyplot as plt
```

```python
# categories
categories = ['infected', 'notinfected']
image_size =180  # Resize images to 64x64

# Load and preprocess images
data_array = []
target_array = []

for i in categories:
    path = os.path.join('Images/train', i)
    class_num = categories.index(i)

    for image_name in os.listdir(path):
        try:
            if image_name.lower().endswith(('.png', '.jpg', '.jpeg')): #to avoid hidden files
                image = cv2.imread(os.path.join(path, image_name), cv2.IMREAD_GRAYSCALE)
                if image is not None:
                    image_resize = cv2.resize(image, (image_size, image_size))
                    data_array.append(image_resize.flatten())  # Flatten to 1D
                    target_array.append(class_num)
        except:
            continue
```

```python
# Convert to NumPy arrays
x = np.array(data_array) / 255.0  # Normalize pixel values to [0, 1]
y = np.array(target_array)

# Split into training and test sets
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
# Train the SVM model
model = SVC(kernel='linear', probability=True)  # Linear kernel
model.fit(x_train, y_train)
```

```python
# Testing the model using the testing data 
y_pred = model.predict(x_test) 

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
print("\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=categories))
```

```python
# Calculating the accuracy of the model 
accuracy = accuracy_score(y_pred, y_test) 
  
# Print the accuracy of the model 
print(f"The model is {accuracy*100}% accurate")
```

```python
path='Sample/notinfected1.jpeg'
image=imread(path) 
plt.imshow(image) 
plt.show() 
image_resize=resize(image,(180,180,1)) 
l=[image_resize.flatten()] 
probability=model.predict_proba(l) 
for ind,val in enumerate(categories): 
	print(f'{val} = {probability[0][ind]*100}%') 
print("The predicted image is : "+categories[model.predict(l)[0]])
```

```python
# 4. Train KNN model
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(x_train, y_train)

# 5. Evaluate performance
y_pred = knn.predict(x_test)

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
print("\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=categories))
```

```python
# Load and preprocess the image
path = 'Sample/infected.jpg'  # Replace with your image path
image = imread(path)
plt.imshow(image, cmap='gray')
plt.axis('off')
plt.title("Test Image")
plt.show()

# Resize and flatten
image_resize = resize(image, (180, 180,1), anti_aliasing=True)
image_flat = image_resize.flatten().reshape(1, -1)  # Reshape for prediction
image_flat = image_flat / 255.0  # Normalize just in case

# Predict with KNN
prediction = knn.predict(image_flat)
print("Predicted class:", categories[prediction[0]])
```
