```python
import os
from PIL import Image
from skimage.io import imread
from skimage.transform import resize
import numpy as np

train_dir = "/Users/tracyojiessy/Library/CloudStorage/OneDrive-UniversityofKeele/AI & Data Masters/Applications of AI, machine learning and data science/ASSESMENT/Practice/data/train" 
```

```python
# List all items in the train folder
PCOS1 = os.listdir(train_dir)
```

```python
# Create a dictionary to map folder names to numbers
PCOS_dict = {}
for i, PCOS in enumerate(PCOS1):
    PCOS_path = os.path.join(train_dir, PCOS)
    if os.path.isdir(PCOS_path):
        PCOS_dict[PCOS] = i
```

```python
# Step 1: Determine minimum image size (skipping broken images)
image_sizes = []
for PCOS in PCOS1:
    PCOS_dir = os.path.join(train_dir, PCOS)
    if not os.path.isdir(PCOS_dir):
        continue
    for image_name in os.listdir(PCOS_dir):
        image_path = os.path.join(PCOS_dir, image_name)
        try:
            image = Image.open(image_path)
            if image.size == (1, 1):
                continue
            image_sizes.append(image.size)
        except Exception as e:
            print(f"Skipping {image_path}: {e}")
```

```python
# Set target size for CNN
image_size = (224, 224)
```

```python
# Step 2: Load and resize all valid images
X = []
y = []

for PCOS in PCOS1:
    fruit_dir = os.path.join(train_dir, PCOS)
    if not os.path.isdir(PCOS_dir):
        continue
    for image_name in os.listdir(PCOS_dir):
        image_path = os.path.join(PCOS_dir, image_name)
        try:
            image = Image.open(image_path)
            if image.size == (1, 1):
                continue
            image = imread(image_path)
            image = resize(image, image_size, anti_aliasing=True)
            X.append(image)
            y.append(PCOS_dict[PCOS])
        except Exception as e:
            print(f"Error loading {image_path}: {e}")
```
